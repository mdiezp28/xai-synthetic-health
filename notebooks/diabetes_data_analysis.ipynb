# Data Analysis: Diabetes Health Indicators Dataset
The Diabetes Health Indicators dataset is a clean dataset of 253,680 survey responses to the CDC's BRFSS2015. The target variable Diabetes_012 has 3 classes. 0 is for no diabetes or only during pregnancy, 1 is for prediabetes, and 2 is for diabetes. There is a class imbalance in this dataset. This dataset has 21 feature variables. https://www.kaggle.com/datasets/alexteboul/diabetes-health-indicators-dataset/data
