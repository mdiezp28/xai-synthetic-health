# Description of the dataset:Parkinson's Telemonitoring
The Parkinson's Telemonitoring dataset includes 5,875 instances from 42 individuals with early-stage Parkinson's disease participating in a six-month trial of a telemonitoring device. The dataset features columns for subject number, age, gender, time from baseline recruitment, motor UPDRS score, total UPDRS score, and 16 biomedical voice measures. Each row corresponds to one voice recording, with the main goal of predicting the motor and total UPDRS scores using the voice measures. 
https://archive.ics.uci.edu/dataset/189/parkinsons+telemonitoring

```python
import pandas as pd
```

```python
pk_data = pd.read_csv("parkinsons-telemonitoring/parkinsons_updrs.data")
#"parkinsons-telemonitoring/parkinsons_updrs.names"
```

## ATTRIBUTE INFORMATION:

- subject - Integer that uniquely identifies each subject
- age - Subject age
- sex - Subject gender '0' - male, '1' - female
- test_time - Time since recruitment into the trial. The integer part is the number of days since recruitment.
- motor_UPDRS - Clinician's motor UPDRS score, linearly interpolated
- total_UPDRS - Clinician's total UPDRS score, linearly interpolated
- Jitter(%),Jitter(Abs),Jitter:RAP,Jitter:PPQ5,Jitter:DDP - Several measures of variation in fundamental frequency
- Shimmer,Shimmer(dB),Shimmer:APQ3,Shimmer:APQ5,Shimmer:APQ11,Shimmer:DDA - Several measures of variation in amplitude
- NHR,HNR - Two measures of ratio of noise to tonal components in the voice
- RPDE - A nonlinear dynamical complexity measure
- DFA - Signal fractal scaling exponent
- PPE - A nonlinear measure of fundamental frequency variation 

Data types: Integer, float and binary.
No missing values.

## Distribution analysis

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.set(style="whitegrid")
```

```python
sns.histplot(unique_patients['age'], bins=20, kde=True)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.show()
```

```python
unique_patients.loc[:, 'gender_label'] = unique_patients['sex'].map({0: 'Male', 1: 'Female'})
sns.countplot(data=unique_patients, x='gender_label')
plt.title("Gender Distribution")
plt.show()
```

```python
# Plot histograms for target variables
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.histplot(pk_data['motor_UPDRS'], bins=30, kde=True)
plt.title('Distribution of motor_UPDRS')
plt.xlabel('motor_UPDRS')
plt.ylabel('Frequency')

plt.subplot(1, 2, 2)
sns.histplot(pk_data['total_UPDRS'], bins=30, kde=True)
plt.title('Distribution of total_UPDRS')
plt.xlabel('total_UPDRS')
plt.ylabel('Frequency')
plt.tight_layout()
plt.savefig('updrs_distributions.png')
plt.show()
```

```python
voice_measures = ['Jitter(%)', 'Shimmer', 'NHR', 'HNR', 'RPDE', 'DFA', 'PPE']
plt.figure(figsize=(15, 10))
for i, measure in enumerate(voice_measures, 1):
    plt.subplot(3, 3, i)
    sns.histplot(pk_data[measure], bins=30, kde=True)
    plt.title(f'Distribution of {measure}')
    plt.xlabel(measure)
    plt.ylabel('Frequency')
plt.tight_layout()
plt.savefig('voice_measures_distributions.png')
plt.show()
```

```python
# Compute correlation matrix
corr_matrix = pk_data.corr()

# Plot heatmap of correlations
plt.figure(figsize=(12, 10))
sns.heatmap(corr_matrix, annot=False, cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Correlation Matrix of All Variables')
plt.savefig('correlation_heatmap.png')
plt.show()

```

```python
# Focus on correlations with target variables
print("\n=== Correlations with motor_UPDRS ===")
print(corr_matrix['motor_UPDRS'].sort_values(ascending=False))
print("\n=== Correlations with total_UPDRS ===")
print(corr_matrix['total_UPDRS'].sort_values(ascending=False))
```

```python
# Scatter plots for key correlated variables
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.scatterplot(x='Jitter(%)', y='motor_UPDRS', data=pk_data)
plt.title('Jitter(%) vs motor_UPDRS')

plt.subplot(1, 2, 2)
sns.scatterplot(x='Shimmer', y='motor_UPDRS', data=pk_data)
plt.title('Shimmer vs motor_UPDRS')
plt.tight_layout()
plt.savefig('scatter_plots.png')
plt.show()
```

```python
sample_subjects = [1, 2, 42]  # Example subjects
plt.figure(figsize=(10, 6))
for subject in sample_subjects:
    subject_data = pk_data[pk_data['subject#'] == subject]
    plt.plot(subject_data['test_time'], subject_data['motor_UPDRS'], label=f'Subject {subject} - motor_UPDRS')
    plt.plot(subject_data['test_time'], subject_data['total_UPDRS'], linestyle='--', label=f'Subject {subject} - total_UPDRS')
plt.title('UPDRS Scores Over Time for Selected Subjects')
plt.xlabel('Test Time (days)')
plt.ylabel('UPDRS Score')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.savefig('longitudinal_trends.png')
plt.show()
```
